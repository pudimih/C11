exercicio 1

```python
musicas = []

while True:
    nome = input("Nome da música: ")
    ano = int(input("Ano da música: "))
    musicas.append({"nome": nome, "ano": ano})

    continuar = input("Deseja cadastrar outra música? [S/N]: ").strip().upper()
    if continuar == "N":
        break

print(f"\nTotal de músicas cadastradas: {len(musicas)}")

ano_mais_antigo = min(m["ano"] for m in musicas)
print("Música(s) do ano mais antigo:")
for m in musicas:
    if m["ano"] == ano_mais_antigo:
        print(f"Nome: {m['nome']} | Ano: {m['ano']}")
```

exercicio 2

```python
import numpy as np

array1 = np.array(["Ana", "Bruno", "Carlos", "Daniel"])
array2 = np.array(["Eduardo", "Fernanda", "Gabriel", "Helena"])

array_concatenado = np.concatenate((array1, array2))

array_2d = array_concatenado.reshape(2, 4)

array_2d_ordenado = np.sort(array_2d.flatten())[::-1].reshape(2, 4)

print(array_2d_ordenado)
```

exercicio 3

```python
import numpy as np

colors = [
    {"color": "black", "type": "primary", "code": {"rgba": [0, 0, 0, 1], "hex": "#000"}},
    {"color": "green", "type": "secondary", "code": {"rgba": [0, 255, 0, 0.1], "hex": "#0F0"}},
    {"color": "yellow", "type": "secondary", "code": {"rgba": [255, 255, 0, 0.7], "hex": "#FF0"}},
    {"color": "blue", "type": "primary", "code": {"rgba": [0, 0, 255, 1], "hex": "#00F"}}
]

print("A)")
for c in colors:
    if c["type"] == "primary":
        print(c["color"])

print("\nB)")
for c in colors:
    if c["code"]["rgba"][2] == 255:
        print(c["code"]["hex"])

print("\nC)")
lista_valores = []
for c in colors:
    lista_valores.append(c["color"])
    lista_valores.append(c["code"]["hex"])
arr_1d = np.array(lista_valores)
print(arr_1d)

print("\nD)")
arr_2d = arr_1d.reshape(-1, 2)
print(arr_2d)

print("\nE)")
traducao = {"black": "preto", "green": "verde", "yellow": "amarelo", "blue": "azul"}
for i in range(len(arr_2d)):
    cor_ingles = arr_2d[i, 0]
    if cor_ingles in traducao:
        arr_2d[i, 0] = traducao[cor_ingles]

print(arr_2d)
```
