```python
seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})
```

```python

#2
print("Total Ano 1:", np.sum(seriesAno1))
print("Total Ano 2:", np.sum(seriesAno2))

```

```python
#3
crescimento = seriesAno2 - seriesAno1
print(crescimento)
```

```python

#4
print(crescimento[crescimento > 0])

```

```python
#5
projecao = seriesAno2 + (crescimento * 2)
print("Linguagem mais popular na projeção:")
print(projecao.nlargest(1))
```

```python
np.random.seed(10)
index = ['A', 'B', 'C', 'D', 'E']
columns = ['W', 'X', 'Y', 'Z']
df = pd.DataFrame(data=np.random.randint(1, 50, [5, 4]), index=index, columns=columns)

#6
colX = df['X']
print("Média (X < 30):", np.mean(colX[colX < 30]))

```

```python
# 7
print("Média da linha D (loc):", np.mean(df.loc['D']))

print("Soma da linha E (iloc):", np.sum(df.iloc[4]))

```

```python
#8
slicing_df = df.loc[['A', 'C', 'E'], ['X', 'Y']]
print("Matriz resultante do Slicing:")
print(slicing_df, "\n")

print("A:", np.sum(slicing_df.loc['A']))
print("C:", np.sum(slicing_df.loc['C']))
print("E:", np.sum(slicing_df.loc['E']), "\n")

print("X:", np.sum(slicing_df['X']))
print("Y:", np.sum(slicing_df['Y']))
```
