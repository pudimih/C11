```python
import pandas as pd
import numpy as np

dfPaises = pd.read_csv('paises.csv', delimiter=';', decimal=',')

dfPaises['Literacy (%)'] = dfPaises['Literacy (%)'].astype(str).str.replace(',', '.').astype(float)
dfPaises['Deathrate'] = dfPaises['Deathrate'].astype(str).str.replace(',', '.').astype(float)
dfPaises['Infant mortality (per 1000 births)'] = dfPaises['Infant mortality (per 1000 births)'].astype(str).str.replace(',', '.').astype(float)
dfPaises['Coastline (coast/area ratio)'] = dfPaises['Coastline (coast/area ratio)'].astype(str).str.replace(',', '.').astype(float)

```

```python
#1 Letra a
paises_oceania = dfPaises[dfPaises['Region'].str.contains('OCEANIA')]
print("Países da Oceania:\n", paises_oceania['Country'])

#1 letra b
print("\nTotal de países da Oceania:", paises_oceania.count()['Country'])

```

```python
#2
maior_populacao = np.max(dfPaises['Population'])
pais_maior = dfPaises[dfPaises['Population'] == maior_populacao]
print(pais_maior[['Country', 'Region']])
```

```python
#3
group_region = dfPaises.groupby('Region')
print(group_region['Literacy (%)'].mean())

```

```python
# 4
noCoast = dfPaises[dfPaises['Coastline (coast/area ratio)'] == 0]
print("Foram encontrados países sem costa e salvos no arquivo.")
noCoast.to_csv('noCoast.csv', sep=';')
```

```python

#5
def analisa_mortalidade(taxa):
    if taxa < 9:
        return 'Balanced'
    else:
        return 'Urgent'

dfPaises['Humanitarian Help'] = dfPaises['Deathrate'].apply(analisa_mortalidade)
print(dfPaises[['Country', 'Deathrate', 'Humanitarian Help']].head())
```

```python
#6
estatisticas_pop = dfPaises.groupby('Region').describe()['Population']
print(estatisticas_pop.head(5))
```

```python
# 7. Função para reduzir mortalidade infantil em 15% e concatenar colunas
def reduz_mortalidade_infantil(x):
    return x * 0.85

infant_original = dfPaises['Infant mortality (per 1000 births)']
infant_reduzido = dfPaises['Infant mortality (per 1000 births)'].apply(reduz_mortalidade_infantil)
infant_reduzido.name = 'Infant mortality (-15%)'

comparacao = pd.concat([infant_original, infant_reduzido], axis=1)
print(comparacao.head())
```

```python
#8
df_sem_coastline = dfPaises.drop('Coastline (coast/area ratio)', axis=1)
df_sem_coastline.to_csv('paises_sem_coastline.csv', sep=';')
print("sucesso!")
```
